```python
import subprocess, sys
# base info
_probe='import torch;' + 'a=torch.zeros(4,device="cuda");b=a+1;torch.cuda.synchronize();print("KERNEL_OK")'
print('--- base torch ---')
r=subprocess.run([sys.executable,'-c',_probe],capture_output=True,text=True,timeout=600)
print('rc',r.returncode,'out',r.stdout.strip(),'err',(r.stderr or '')[-200:])
print('--- try torch 2.5.1 cu121 (no numpy reinstall) ---')
subprocess.check_call([sys.executable,'-m','pip','install','-q','--no-cache-dir','--index-url','https://download.pytorch.org/whl/cu121','torch==2.5.1'],timeout=1800)
r=subprocess.run([sys.executable,'-c',_probe],capture_output=True,text=True,timeout=600)
print('rc',r.returncode,'out',r.stdout.strip(),'err',(r.stderr or '')[-200:])
# now numpy import + pandas after new torch
r=subprocess.run([sys.executable,'-c','import numpy as np, pandas as pd; print(np.__version__, pd.__version__)'],capture_output=True,text=True,timeout=600)
print('np/pd rc',r.returncode,'out',r.stdout.strip(),'err',(r.stderr or '')[-300:])
```
